```python
import lightgbm as lgb
import pandas as pd
```

```python
class Config:
    lgb_params = {
        "boosting_type": "gbdt",
        "objective": "regression",
        "metric": {"l2", "l1"},
        "num_leaves": 31,
        "learning_rate": 0.05,
        "feature_fraction": 0.9,
        "bagging_fraction": 0.8,
        "bagging_freq": 5,
        "verbose": 0,
    }
    num_boost_round = 100
    stopping_rounds = 5

config = Config()
```

```python
df_train = pd.read_csv("data/coral_health_training.csv", dtype="float32")

y_train = df_train["coral_health_index"]
X_train = df_train.drop("coral_health_index",axis=1)
lgb_train = lgb.Dataset(X_train, y_train)
```

```python
gbm = lgb.train(
    config.lgb_params,
    lgb_train,
    num_boost_round=config.num_boost_round,
    # callbacks=[lgb.early_stopping(stopping_rounds=config.stopping_rounds)]
)
```

```python
gbm.save_model("model.txt")
```
